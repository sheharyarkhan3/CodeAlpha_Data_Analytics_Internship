```python
#Task 1: Web Scraping
#CodeAlpha Data Analytics Internship
#In this notebook, I collect a custom dataset of books from a public website using Python.

## Import Libraries
#requests: downloads the web page
#BeautifulSoup: reads the HTML and finds the data I need
#pandas: organizes the data into a table and saves it as CSV

import requests
from bs4 import BeautifulSoup
import pandas as pd

## Objective
#The goal of this task is to extract data from a website, handle its HTML structureand page navigation, and save the result as a clean dataset for later analysis.

base_url = "https://books.toscrape.com/catalogue/page-{}.html"
ratings = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
books = []
## Website Chosen
#URL:https://books.toscrape.com

#(Why this website?
# It is a public website made for practicing web scraping
# It has 50 pages, so I can practice web navigation
# Each book has a title, price, rating, and availability, which gives a useful dataset)
## Scrape the Data
#How the scraper works:
#1. It loops through pages 1 to 50 of the website
#2. For each page, it downloads the HTML and parses it with BeautifulSoup
#3. It finds every book on the page (each one sits inside an `article` tag with class `product_pod`)
#4. It extracts the title, price, star rating, and availability
#5. It stores every book as a dictionary in a list
for page in range(1, 51):  # the site has 50 pages
    response = requests.get(base_url.format(page))
    soup = BeautifulSoup(response.text, "html.parser")

    for item in soup.select("article.product_pod"):
        title = item.h3.a["title"]
        price = item.select_one(".price_color").text.strip()
        rating = ratings[item.select_one("p.star-rating")["class"][1]]
        stock = item.select_one(".availability").text.strip()
        books.append({
            "Title": title,
            "Price": price,
            "Rating": rating,
            "Availability": stock
        })
## Create the Dataset
#I convert the list into a pandas DataFrame and save it as a CSV file.
df = pd.DataFrame(books)
df.to_csv("books.csv", index=False)
print(df.shape)
df.head()

## Results
#Successfully scraped **1000 books** from 50 pages
#The dataset has 4 columns: Title, Price, Rating, Availability
#Saved as `books.csv`
## What I Learned
#How to inspect the HTML structure of a page using Chrome DevTools
#How to use BeautifulSoup to select tags and classes
#How to loop through multiple pages to collect data
# How to turn scraped data into a pandas DataFrame

## Next Steps
#The Price column contains the £ symbol. In the next task, I will clean it and analyze the data.
```
